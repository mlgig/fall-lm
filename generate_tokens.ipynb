```python
%load_ext autoreload
%autoreload 2

import os
import numpy as np
import pandas as pd
import joblib
```

```python
INDEX_FILE = "features/feature_index_w3_s1.csv"

TOKENS_DIR = "tokens/w3_k60_s1"
os.makedirs(TOKENS_DIR, exist_ok=True)
```

```python
# Load kmeans model and scaler
kmeans = joblib.load("motion_vocab_k60.joblib")
scaler = joblib.load("feature_scaler_w3_s1.joblib")

# Load index
index_df = pd.read_csv(INDEX_FILE)
print("Files in index:", len(index_df))
index_df.head()
```

```python
for _, row in index_df.iterrows():
    
    feature_path = row["feature_path"]
    subject_id = row["subject_id"]
    file_id = row["file_id"]
    
    # load features
    feats = np.load(feature_path)

    # scale
    feats_scaled = scaler.transform(feats)

    # predict tokens
    tokens = kmeans.predict(feats_scaled)

    # save
    token_file = f"{subject_id}__{file_id}.npy"
    token_path = os.path.join(TOKENS_DIR, token_file)

    np.save(token_path, tokens)

    print("Saved tokens:", subject_id, file_id)
```
