```python
%load_ext autoreload
%autoreload 2

import os
import numpy as np
import pandas as pd
```

```python
INDEX_FILE = "features/feature_index_w3_s1.csv"

SAX_DIR = "tokens_sax/w3_a5_s1"
os.makedirs(SAX_DIR, exist_ok=True)

ALPHABET_SIZE = 5
```

```python
from scipy.stats import norm

# SAX breakpoints from Gaussian quantiles
def sax_breakpoints(alphabet_size):
    return norm.ppf(np.linspace(0, 1, alphabet_size + 1)[1:-1])

breakpoints = sax_breakpoints(ALPHABET_SIZE)

def sax_encode(values, breakpoints):
    return np.digitize(values, breakpoints)

def window_magnitude_from_features(features):
    # magnitude mean stored as one of the feature columns
    # index depends on feature order — adjust if needed
    # assuming magnitude mean is at position 15
    return features[:, 15]
```

```python
index_df = pd.read_csv(INDEX_FILE)

for _, row in index_df.iterrows():
    
    feature_path = row["feature_path"]
    subject_id = row["subject_id"]
    file_id = row["file_id"]

    feats = np.load(feature_path)

    # get magnitude proxy
    mag_series = window_magnitude_from_features(feats)

    # normalize (SAX requires z-normalisation)
    mag_norm = (mag_series - np.mean(mag_series)) / (np.std(mag_series) + 1e-8)

    # encode
    sax_tokens = sax_encode(mag_norm, breakpoints)

    # save
    token_file = f"{subject_id}__{file_id}.npy"
    token_path = os.path.join(SAX_DIR, token_file)

    np.save(token_path, sax_tokens)

    print("Saved SAX:", subject_id, file_id)
```
